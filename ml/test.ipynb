```python
import pandas as pd
```

```python
ratings = pd.read_csv("/home/sreenandump/Project/semantic-movie-recommender-two-tower/ml/data/raw/ml-latest-small/ratings.csv")
movies = pd.read_csv("/home/sreenandump/Project/semantic-movie-recommender-two-tower/ml/data/raw/ml-latest-small/movies.csv")
```

```python
movie_stats = ratings.groupby("movieId").agg({
    "rating": ["mean", "count"]
})
movie_stats.columns = ["avg_rating", "rating_count"]
movie_stats.head(10)
```

```python
popular = movie_stats.sort_values(
    by=["avg_rating", "rating_count"],
    ascending=False
)
popular.head(10)
```

```python
popular = popular.merge(
    movies,
    on="movieId"
)
popular.head(10)
```
